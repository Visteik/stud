```python
import matplotlib.pyplot as plt
import numpy as np

# Данные
days = list(range(1, 15))
day_temp = [5.1, 6.3, 4.8, 7.2, 9.5, 8.1, 6.7, 5.9, 7.4, 8.8, 10.2, 9.1, 7.5, 6.2]
night_temp = [-1.2, 0.3, -2.1, 1.5, 3.8, 2.4, 0.9, -0.4, 1.8, 3.2, 4.6, 2.9, 1.1, -0.3]

# Вычисления
avg_day = np.mean(day_temp)
avg_night = np.mean(night_temp)
cold_nights = sum(1 for t in night_temp if t < 0)

print(f"Средняя температура днём: {avg_day:.1f}°C")
print(f"Средняя температура ночью: {avg_night:.1f}°C")
print(f"Ночей с температурой ниже 0: {cold_nights}")

# Построение графиков
plt.figure(figsize=(10, 5))

# Днём: сплошная линия с маркерами 'o'
plt.plot(days, day_temp, marker='o', linestyle='-', color='orange', label='Днём')

# Ночью: штриховая линия с маркерами 's'
plt.plot(days, night_temp, marker='s', linestyle='--', color='blue', label='Ночью')

# Горизонтальные линии через plot (от 1 до 14 по X)
plt.plot([1, 14], [avg_day, avg_day], color='orange', linestyle=':', label='Среднее днём')
plt.plot([1, 14], [avg_night, avg_night], color='blue', linestyle=':', label='Среднее ночью')
plt.plot([1, 14], [0, 0], color='red', linestyle='-', label='0°C')

# Оформление
plt.title('Температура воздуха за 2 недели')
plt.xlabel('День')
plt.ylabel('Температура (°C)')
plt.xticks(days)
plt.grid(True)
plt.legend()

# Сохранение и показ
plt.savefig('weather.png')
plt.show()
```

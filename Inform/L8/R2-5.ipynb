```python
import matplotlib.pyplot as plt
import numpy as np

# Данные
days = list(range(1, 15))
display = [4.5,	5.2, 5.8, 4.9, 6.1,	6.8, 7.2, 7.5, 8.1,	7.8, 8.5, 9.0, 9.3,	9.7]
battery = [58, 52, 48, 55, 42, 38, 33, 29, 22, 25, 18, 14, 11, 8]

# Вычисления
avg_display = np.mean(display)
avg_battery = np.mean(battery)


print(f"Средняя экранного времени': {avg_display:.1f} ч")
print(f"Средняя заряда баттареи: {avg_battery:.1f}%")


# Построение графиков
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 6))

# Левая панель: график экранного времени
ax1.plot(days, display, marker='o', color='blue', label='Экранное время')
ax1.set_title('Экранное время')
ax1.set_xlabel('День')
ax1.set_ylabel('Часы')

# Линия среднего значения для левой панели
ax1.axhline(avg_display, color='red', linestyle='--', label=f'Среднее: {avg_display:.2f}')
ax1.legend()

# Правая панель: график заряда батареи
ax2.plot(days, battery, marker='s', color='green', label='Заряд батареи')
ax2.set_title('Заряд батареи')
ax2.set_xlabel('День')
ax2.set_ylabel('Процент (%)')
# Линия среднего значения для правой панели
ax2.axhline(avg_battery, color='red', linestyle='--', label=f'Среднее: {avg_battery:.2f}')
ax2.legend()

# Настроить пределы осей
ax1.set_ylim(0, 11)
ax2.set_ylim(0, 100)

# Сохранение и показ
plt.savefig('phone.png')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Запись массивов
names = ["Маргарита Мания", "Пепперони-Парад", "Гавайский Хаос", "Четыре Сыра и Слёзы", "Веганский Бунт"]
sales = [23, 45, 56, 78, 32]

# Числовой массив позиций
x = np.arange(len(names))

# Поиск победителя
winner_idx = np.argmax(sales)

# Создание фигуры и осей
fig, ax = plt.subplots(figsize=(10, 5))

# Чтобы выделить победителя созадем массив цветов и красим все в синий, 
# а потом нужный нам элемент красим в золотой

colors = ['steelblue']*len(names)
colors[winner_idx] = 'gold'

# Диаграмма
bars = ax.bar(x, sales, width=0.6, color=colors, edgecolor='navy')

# Оси
ax.set_title("Битва пиццерий: продажи за неделю")
ax.set_xlabel("Пиццерия")
ax.set_ylabel("Продано пицц шт.")

ax.set_xticks(x)
ax.set_xticklabels(names)
ax.bar_label(bars)

plt.savefig('pizzeria.png')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Запись массивов
labels = ["Спасает мир", "Восстанавливает супер-силы во сне", "отдыхает, играя в кибер-квесты", "занимается «другим» \n(тайные миссии, а возможно,\n поедание доширака)"]
hours = [8, 7, 5, 4]

fig, ax = plt.subplots(figsize=(10, 7))

ax.pie(hours, labels=labels, autopct='%1.1f%%')

ax.set_title("День супергероя Алексея")

plt.savefig('superhero.png', dpi=150)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Запись массивов
dist = np.array([100, 500, 1000, 3000, 5000, 10000])
ping = np.array([1.8, 5.9, 11.5, 32.5, 53.0, 104.2])


c = 300000  # Скорость света в вакууме, км/с
n = 1.5     # Коэффициент преломления
v = c / n   # Скорость света в оптоволокне, км/с

ping_theory = (2 * dist / v) * 1000

fig, ax = plt.subplots(figsize=(10, 20))

ax.scatter(dist, ping, linestyle='None', marker = 'o', color = 'red', label='Измеренный пинг')

ax.plot(dist, ping_theory, linestyle='-', color='navy', label='Теоретический пинг')

# np.concatenate - склеивает 2 массива
# np.unique - избавляется от повторов в массиве и выстраивает его по возрастанию
all_ping = np.unique(np.concatenate((ping, ping_theory)))

ax.set_xticks(dist)  
ax.set_yticks(all_ping)

ax.set_title("Зависимость пинга от расстояния в оптоволокне")
ax.set_xlabel("Расстояние, км")
ax.set_ylabel("Пинг, мс")
ax.grid(True, linestyle = '--', alpha=0.7)
ax.legend()

plt.savefig('ping_test.png', dpi=150)
plt.show()
```

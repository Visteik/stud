```python
import matplotlib.pyplot as plt
import numpy as np

# Данные
days = list(range(1, 15))
display = [4.5,	5.2, 5.8, 4.9, 6.1,	6.8, 7.2, 7.5, 8.1,	7.8, 8.5, 9.0, 9.3,	9.7]
battery = [58, 52, 48, 55, 42, 38, 33, 29, 22, 25, 18, 14, 11, 8]

# Вычисления
avg_display = np.mean(display)
avg_battery = np.mean(battery)


print(f"Средняя экранного времени': {avg_display:.1f} ч")
print(f"Средняя заряда баттареи: {avg_battery:.1f}%")


# Построение графиков
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 6))

# 2. Левая панель: график экранного времени
ax1.plot(days, display, marker='o', color='blue', label='Экранное время')
ax1.set_title('Экранное время')
ax1.set_xlabel('День')
ax1.set_ylabel('Часы')

# 3. Линия среднего значения для левой панели
ax1.axhline(avg_display, color='red', linestyle='--', label=f'Среднее: {avg_display:.2f}')
ax1.legend()

# 2. Правая панель: график заряда батареи
ax2.plot(days, battery, marker='s', color='green', label='Заряд батареи')
ax2.set_title('Заряд батареи')
ax2.set_xlabel('День')
ax2.set_ylabel('Процент (%)')
# 3. Линия среднего значения для правой панели
ax2.axhline(avg_battery, color='red', linestyle='--', label=f'Среднее: {avg_battery:.2f}')
ax2.legend()

# 4. Настроить пределы осей
ax1.set_ylim(0, 11)
ax2.set_ylim(0, 100)

# Сохранение и показ
plt.savefig('phone.png')
plt.show()
```

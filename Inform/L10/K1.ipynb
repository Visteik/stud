```python
with open("files/server_log.txt", "r", encoding="utf-8") as log:
    lines = log.readlines()


error_lines = []
count = 0
for line in lines:
    if "ERROR" in line or "CRITICAL" in line:
        error_lines.append(line)
        count += 1

with open("files/errors_only.txt", "w", encoding="utf-8") as er:
    er.writelines(error_lines)
    

print("Найдено", count, "критических ошибок")
```

```python
error_count = {"ERROR": 0,
               "INFO": 0,
                "WARNING": 0,
                "CRITICAL": 0
               }


with open("files/server_log.txt", "r", encoding="utf-8") as log:
    for line in log:
        
        line = line.strip()
        part = line.split()

        error = part[2]
        if error in error_count:
            error_count[error] += 1

for er, count in error_count.items():
    print(f"{er}: {count}")
```

```python
with open("files/secret.txt", "r", encoding="utf-8") as log:
    lines = log.readlines()


res_line = []
for line in lines:
    parts = line.strip().split(",")
    
    new_parts = []
    for part in parts:
        part = part.strip()
        if part.startswith("тел:"):
            phone = part.replace("тел:", "").strip()
            mask_phone = "*"*(len(phone)-2)+phone[-2:]
            new_parts.append(f"тел: {mask_phone}")
        elif part.startswith("email:"):
            mail = part.replace("email:", "").strip()
            idx = mail.find('@')
            
            a_mail = mail[idx:]

            mask_mail = mail[0]+'***'+a_mail
            new_parts.append(f"email: {mask_mail}")
        else:
            new_parts.append(part)    
    res_line.append(", ".join(new_parts) + "\n")


with open("files/secret_anonymized.txt", "w", encoding="utf-8") as out:
    out.writelines("".join(res_line))
```

```python
import string


with open("files/article.txt", "r", encoding="utf-8") as log:
    line = log.readline()

line = line.lower()
text = ''.join(ch for ch in line if ch not in string.punctuation)

words = text.split()
words_count = {}
for word in words:
    if word in words_count:
        words_count[word] += 1
    else:
        words_count[word] = 1
top_words = sorted(words_count.items(), key=lambda x: x[1], reverse=True)
for word, count in top_words[:5]:
    print(f"{word}: {count}")
```

```python
import string


def t9(text):
    out_text = {}
    symb_code = {
        "2": "abc",
        "3": "def",
        "4": "ghi",
        "5": "jkl",
        "6": "mno",
        "7": "pqrs",
        "8": "tuv",
        "9": "wxyz",
        "0": " "
        }
    
    for key, symb in symb_code.items():
        for i, ch in enumerate(symb, start=1):
            out_text[ch] = key * i
            
    result = []
    for char in text.lower():
        result.append(out_text.get(char, ''))

    return "".join(result)

print(t9("ARF wfsfw aw"))
```

```python
import json

with open("files/playlist.json", "r", encoding="utf-8") as f:
    loaded_data = json.load(f)

print("Поиска жанра 1. Добавить трек 2")
t = input()

#Вывод доступных жанров
print('Достпуные жанры: ')
for genre in loaded_data.keys():
    print(genre)

match t:
    case "1":
        #Поиск жанра
        user_genre = input('Какой жанр ищешь: ').strip().capitalize()
        if user_genre in loaded_data:
            print('Доступные треки:')
            for track in loaded_data[user_genre]:
                print(track)
        else:
            print('Такого жанра нет')
    case '2':
        #Добавить трек
        user_track = input('Введите трек для добавления').strip().capitalize()
        user_gen = input('Введите жанр трека').strip().capitalize()

        if user_gen in loaded_data:
            loaded_data[user_gen].append(user_track)
            print(f"Трек добавлен в жанр {user_gen}!")
        else:
            loaded_data[user_gen] = [user_track]
            print(f"Создан новый жанр {user_gen} и добавлен трек")

        with open('files/play.json', 'w', encoding='utf-8') as file:
            json.dump(loaded_data, file, ensure_ascii=False, indent=4)
```
